```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

### Matplot Basics


```python
x = np.linspace(0,5,11)
x
y = x**3
y
```

```python
plt.plot(x,y)
plt.title("Hello")
plt.xlabel("X-axis")
plt.ylabel('Y-axis')
```

### Subplots in Matplot

```python
plt.subplot(2,2,1)
plt.plot(y,x)
plt.subplot(2,2,2)
plt.plot(y,x)
plt.subplot(2,2,3)
plt.plot(-y,-y)
plt.subplot(2,2,4)
plt.plot(x,x)
```

```python
plt.subplot(6,6,1)
plt.subplot(6,6,2)
plt.subplot(6,6,3)
plt.subplot(6,6,4)
plt.subplot(6,6,5)
plt.subplot(6,6,6)
```

### Object Oreanted way
#### We have full control of our canvas

Matplotlib allows the aspect ratio, DPI and figure size to be specified when the Figure object is created. You can use the figsize and dpi keyword arguments.

figsize is a tuple of the width and height of the figure in inches dpi is the dots-per-inch (pixel per inch).

```python
fig = plt.figure(figsize=(10,5), dpi = 200)
axis1 = fig.add_axes([0.1,0.1,1,1])
axis1.plot(x,y)
axis2 = fig.add_axes([0.35,0.35,0.5,0.5])
axis2.plot(x,y)
```
