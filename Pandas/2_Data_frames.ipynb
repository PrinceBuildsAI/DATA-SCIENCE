# Data Frames 
### (Data Frames = combination of multiple series)

#### Creating a Data Frame

```python
import numpy as np
import pandas as pd
```

```python
data = {
    'Name': ['John', 'Anna', 'Peter', 'Linda'],
    'Age': [28, 34, 29, 42],
    'City': ['New York', 'Paris', 'Berlin', 'London'],
    'Salary': [65000, 70000, 62000, 85000]
}
df = pd.DataFrame(data)
```

```python
df
```

```python
data_list = [
    ['John', 28, 'New York', 65000],
    ['Anna', 34, 'Paris', 70000],
    ['Peter', 29, 'Berlin', 62000],
    ['Linda', 42, 'London', 85000]
]
df2 = pd.DataFrame(data_list)
columns = ["Name", "Age", "City", "salary"]
df2 = pd.DataFrame(data_list, columns= columns)
df2
```

#### Selecting and indexing of columns

```python
df2
```

```python
df2['Name']
```

```python
df2[['Name', 'City']]
```

#### Creating a new column

```python
df2["Designation"] = ["Doctor", "Engineer", "Teacher", "Doctor"]
df2
```

```python
df2.drop("Designation", axis = 1)   #only removes column in secondary storage
```

```python
df2.drop("Designation", axis = 1, inplace= True)    #Inplace removes data from main data farame
```

#### Conditional Selection

```python
df2
```
