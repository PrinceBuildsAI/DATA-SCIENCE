# **Array indexing**

```python
import numpy as np
```

#### NUMPY INDEXING AND SLICING OF VECTORS

```python
arr = np.arange(10,21)
arr
```

```python
arr[6]  #accessing the element
```

```python
arr[1:6]    #slicing is just like list in python
```

```python
arr[3::2]
```

#### NUMPY INDEXING AND SLICING OF A MATRIX 

```python
arr = np.arange(1,31).reshape(6,5)
arr
```

```python
arr[0]  # singel number will target rows of matrix
```

```python
arr[:,0:1] #Row(all),column(one)
```

```python
arr[3,0]    #arr[row,column] #3rd row 0th column
```

```python
arr[0:2,1:4] #slicing of Matrix
```

```python
arr[3:,3:]
```

#### BOOLEAN INDEXING

```python
arr = np.arange(11,21)
arr
```

```python
bool_index = arr % 2 == 0
bool_index
```

```python
arr = arr[bool_index]
arr
```
