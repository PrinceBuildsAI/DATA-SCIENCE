#### Arithmatic operations

```python
a1 = np.array([1,2,3,4,5])
a2 = np.array([6,7,8,9,10])
```

```python
arr2 = a1+a2    #Number of elements nust be same in both array 
arr3 = a1-a2
arr4 = a1*a2
arr5 = a1/a2
arr6 = a1**a2
print(arr2)
print(arr3)
print(arr4)
print(arr5)
print(arr6)
```

## Advance Array Manupulation

#### Stacking Array

```python
a = np.array([1,2,3,4])     # stacking of array means combining of array and create a new one
b = np.array([5,6,7,8])
```

```python
np.stack((a,b))
```

```python
np.vstack((a,b))    #verticle-stack
```

```python
np.hstack((a,b))    #horizontal-stack
```

```python
np.column_stack((a,b))  # column_stack converts 1d array into 2d array
```
