#### Array Generation Function

```python
arr = np.arange(1,11,2)   # range (1 to 10 with 2 gap)
arr
```

```python
arr = np.zeros(6)
arr
```

```python
arr = np.zeros((6,6)) #matrix to zeroes [6,6]
arr
```

```python
arr = np.linspace(1,5,10) #Provides equal distance element between generated elements
arr
```

#### Random Generation Function

```python
np.random.rand(10) # Normalization (When all the values are comes between 0 and 1) statistics
```

```python
np.random.randn(10) #Standardization (When ramdom number comes between -3 to 3)
```

```python
np.random.randint(10,20,10) # first two (10,20) range 10 is limit 
```
